```python
# Task 9: Python Data Processing

# Step 1: Create a collection of student records
students = [
    {"name": "Vasundhara", "age": 20, "marks": 85, "department": "CSE"},
    {"name": "Akshitha", "age": 21, "marks": 90, "department": "ISE"},
    {"name": "Thanushree", "age": 20, "marks": 78, "department": "CSE"},
    {"name": "Rahul", "age": 22, "marks": 65, "department": "ECE"},
    {"name": "Priya", "age": 21, "marks": 95, "department": "CSE"}
]


# Step 2: Display all student records
def display_records(records):
    print("STUDENT RECORDS")
    print("-" * 50)

    for student in records:
        print(
            f"Name: {student['name']}, "
            f"Age: {student['age']}, "
            f"Marks: {student['marks']}, "
            f"Department: {student['department']}"
        )


# Step 3: Calculate summary statistics
def calculate_statistics(records):
    if not records:
        print("No records available.")
        return

    marks = [student["marks"] for student in records]

    total_students = len(records)
    total_marks = sum(marks)
    average_marks = total_marks / total_students
    highest_marks = max(marks)
    lowest_marks = min(marks)

    passed_students = [
        student for student in records
        if student["marks"] >= 40
    ]

    failed_students = [
        student for student in records
        if student["marks"] < 40
    ]

    print("\nSUMMARY STATISTICS")
    print("-" * 50)
    print("Total Students:", total_students)
    print("Total Marks:", total_marks)
    print("Average Marks:", round(average_marks, 2))
    print("Highest Marks:", highest_marks)
    print("Lowest Marks:", lowest_marks)
    print("Passed Students:", len(passed_students))
    print("Failed Students:", len(failed_students))


# Step 4: Find students scoring above a given mark
def filter_students(records, minimum_marks):
    filtered_records = [
        student for student in records
        if student["marks"] >= minimum_marks
    ]

    print(f"\nStudents scoring {minimum_marks} or above:")

    for student in filtered_records:
        print(student["name"], "-", student["marks"])


# Step 5: Find the topper
def find_topper(records):
    if not records:
        print("No records available.")
        return

    topper = max(records, key=lambda student: student["marks"])

    print("\nTOPPER")
    print("Name:", topper["name"])
    print("Marks:", topper["marks"])
    print("Department:", topper["department"])


# Step 6: Execute the program
display_records(students)
calculate_statistics(students)
filter_students(students, 80)
find_topper(students)
```
